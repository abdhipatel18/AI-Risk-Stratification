This notebook demonstrates exploratory analysis, clinical feature distribution, rules-based risk scoring, and machine learning classifier comparison for patient scheduling prioritization.

```python
import sys
import os
sys.path.insert(0, os.path.abspath(".."))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.main import run_pipeline

# Run pipeline to generate analytical dataset
df_processed, df_quality, ml_summary = run_pipeline(data_dir="../data/input", config_dir="../config", output_dir="../data/output")
df_processed.head()
```

### 1. Risk Score Distribution
Visualize overall patient risk score distribution and risk category breakdown.

```python
plt.figure(figsize=(10, 5))
sns.histplot(df_processed['calculated_risk_score'], kde=True, bins=20, color='navy')
plt.title('Distribution of Patient Calculated Risk Scores')
plt.xlabel('Risk Score')
plt.ylabel('Patient Count')
plt.show()
```
